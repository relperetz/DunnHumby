# Git and Virtual Environment

This notebook summarizes the Git workflow and Python virtual environment setup used in the Dunnhumby project.

The goal is to understand:
- How Git tracks project changes
- How GitHub is connected to the local project
- How to create and use a Python virtual environment
- How to install and manage project dependencies
- How to keep the project environment isolated and reproducible

## 1. Git

Git is a version control system.

It allows us to:
- Track changes in project files
- Save versions of the project
- Return to previous versions
- Work safely without losing previous work
- Connect the local project to GitHub

Git works locally on the computer. GitHub is an online service where the Git repository can be stored and shared.

## 2. Creating a Git Repository

### `git init`

```powershell
git init
```

Turns a normal folder into a Git repository. It creates the hidden `.git` folder that stores Git history and configuration.

**Project note:** The Dunnhumby project already had a Git repository, so we did not need to run `git init` again.

## 3. Checking Git Status

### `git status`

```powershell
git status
```

Shows the current state of the project in Git, including modified, untracked, staged, and committed files.

This is one of the most useful Git commands and should be used frequently.

## 4. Staging Changes

### Add all changes

```powershell
git add .
```

Adds all current changes under the current folder to the staging area.

### Add one specific file

```powershell
git add README.md
```

The staging area is the place where we select which changes will be included in the next commit.

Workflow:

```text
Files → git add → STAGING → git commit → Local Git → git push → GitHub
```

## 5. Creating a Commit

### `git commit`

```powershell
git commit -m "Add data quality analysis"
```

Creates a saved point in the Git history.

The text after `-m` is a short description of the change.

## 6. Viewing Git History

```powershell
git log
```

Shows the Git commit history.

A shorter version is:

```powershell
git log --oneline
```

The short code shown for each commit is its commit identifier.

## 7. GitHub and Remote Repositories

### `git remote -v`

```powershell
git remote -v
```

Shows which remote repository the local project is connected to.

The conventional name for the main remote is `origin`.

In this project the remote is the Dunnhumby GitHub repository.

## 8. Push – Sending Changes to GitHub

```powershell
git push
```

Sends local commits to GitHub.

For the first push of a branch, we may use:

```powershell
git push -u origin main
```

The `-u` option establishes the upstream relationship between the local branch and the remote branch. After that, `git push` is usually enough.

## 9. Fetch and Pull

### `git fetch`

```powershell
git fetch
```

Downloads information about changes from GitHub without integrating those changes into the working files.

### `git pull`

```powershell
git pull
```

Gets changes from GitHub and integrates them into the local project.

Conceptually:

```text
git pull ≈ git fetch + integration
```

## 10. Discarding an Unwanted Local Change

### `git restore`

```powershell
git restore <filename>
```

Example:

```powershell
git restore 02_notebooks/01_data_loading.ipynb
```

Restores the file to the version currently recorded by Git and discards the local uncommitted changes.

**Important:** This can remove work that has not been committed.

## 11. `.gitignore`

The `.gitignore` file tells Git which files and folders should not be tracked.

Examples used in this project:

```gitignore
.venv/
01_data/01_raw/
__pycache__/
.ipynb_checkpoints/
.env
```

This keeps the repository clean and prevents large raw datasets, the virtual environment, temporary files, and secrets from being committed.

## 12. Python Virtual Environment

A virtual environment creates an isolated Python environment for a specific project.

Instead of installing project packages into the global Python installation, packages are installed inside the project's `.venv` folder.

This prevents different projects from interfering with each other.

Project environment:

```text
C:\Projects\DunnHumby\.venv
```

## 13. Creating the Virtual Environment

```powershell
python -m venv .venv
```

Explanation:
- `python` runs Python
- `-m` tells Python to run a module
- `venv` is the module used to create virtual environments
- `.venv` is the folder where the environment is created

## 14. Activating the Virtual Environment

In PowerShell:

```powershell
.venv\Scripts\Activate.ps1
```

When the environment is active, the terminal normally shows `(.venv)` at the beginning of the prompt.

Example:

```text
(.venv) PS C:\Projects\DunnHumby>
```

## 15. Using the Virtual Environment Python Directly

The Python executable inside the environment can be called directly:

```powershell
.venv\Scripts\python.exe
```

For example:

```powershell
.venv\Scripts\python.exe -m pip install -r requirements.txt
```

This is useful because it guarantees that the command uses the project's Python interpreter.

## 16. Why `python -m pip`?

Instead of:

```powershell
pip install pandas
```

we can use:

```powershell
python -m pip install pandas
```

This means: use this Python interpreter to run the `pip` module.

The most explicit version for this project is:

```powershell
.venv\Scripts\python.exe -m pip install pandas
```

This removes ambiguity about which Python installation is being used.

## 17. Installing Dependencies from `requirements.txt`

```powershell
.venv\Scripts\python.exe -m pip install -r requirements.txt
```

The `-r` option tells pip to read the package list from the specified file.

`requirements.txt` makes it possible to recreate the project's dependencies on another computer.

## 18. Checking Installed Packages

```powershell
.venv\Scripts\python.exe -m pip list
```

Shows installed packages and their versions.

Another useful command is:

```powershell
.venv\Scripts\python.exe -m pip freeze
```

This outputs package versions in a format commonly used for dependency files.

## 19. Creating a Jupyter Kernel

```powershell
.venv\Scripts\python.exe -m ipykernel install --user --name dunn-humby --display-name "Python (DunnHumby)"
```

This registers the project's virtual environment as a Jupyter kernel.

The kernel can then be selected in VS Code/Jupyter as:

`Python (DunnHumby)`

This ensures that notebooks use the project's virtual environment.


Project Python:

The global Python can be used by other projects. The `.venv` Python is isolated for Dunnhumby.

## 24. Why `.venv` Is Not Uploaded to GitHub

The `.venv` folder contains many files specific to the local computer and operating environment.

There is no need to store it in GitHub.

Instead, the project stores `requirements.txt`.

On another computer we can create a new environment and reinstall the required packages:

```powershell
python -m venv .venv
.venv\Scripts\python.exe -m pip install -r requirements.txt
```

## 26. The Big Picture

The project uses two separate systems:

**Git**
- Tracks project files and their history
- Synchronizes the project with GitHub

**Virtual Environment**
- Isolates Python and project dependencies
- Keeps project packages separate from global Python

Together they make the project organized, reproducible, and easier to maintain.
