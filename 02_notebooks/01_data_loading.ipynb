# Dunnhumby Retail Analytics

## 1. Project Introduction

This notebook is the first stage of the retail analytics project.

The goal of this stage is to load the raw Dunnhumby datasets and understand
their structure before performing any data cleaning or transformation.

### Objectives

- Identify all available datasets
- Understand the structure of each dataset
- Inspect data types and missing values
- Identify the size of each dataset
- Understand the relationships between datasets

```python
from pathlib import Path
import pandas as pd

# Load all CSV files from the raw data folder
RAW_DIR = Path("../01_data/01_raw")

# Find all CSV files
csv_files = sorted(RAW_DIR.glob("*.csv"))

print(f"Found {len(csv_files)} CSV files:")
for file in csv_files:
    print(f" - {file.name}")
```

## 2. Load the Raw Datasets

The project contains several datasets covering different aspects of the
retail environment.

These datasets include:

- Customer transactions
- Products
- Coupons
- Campaigns
- Customer demographics
- Promotional activity

At this stage, the data is loaded as-is.

No cleaning or transformation is performed yet.

```python
datasets = {}

for file in csv_files:
    df = pd.read_csv(file)
    datasets[file.stem] = df
    
    print(
        f"{file.name}: "
        f"{df.shape[0]:,} rows × {df.shape[1]} columns"
    )
```

## 3. Inspect the Dataset Contents

The next step is to look at the first rows of each dataset.

This helps us understand:

- What information each dataset contains
- How the columns are named
- Whether values appear as expected
- Which datasets may later need to be joined

```python
for name, df in datasets.items():
    print(f"\n===== {name} =====")
    display(df.head())
```

## 4. Dataset Overview

Before analyzing individual datasets, we will create a high-level summary.

The summary will help us understand:

- The size of each dataset
- The number of columns
- The number of missing values
- The number of duplicate rows

This gives us an initial understanding of the data quality and helps identify
datasets that may require special attention in the next stage.

```python
dataset_summary = pd.DataFrame([
    {
        "dataset": name,
        "rows": len(df),
        "columns": df.shape[1],
        "missing_values": int(df.isna().sum().sum()),
        "duplicate_rows": int(df.duplicated().sum())
    }
    for name, df in datasets.items()
])

dataset_summary = dataset_summary.sort_values(
    "rows",
    ascending=False
).reset_index(drop=True)

display(dataset_summary)
```

## 5. Column Structure

The next step is to inspect the columns available in each dataset.

Understanding the column structure is important because the datasets are
related to each other through identifiers such as household, product,
campaign, store, and basket.

At this stage, we are only inspecting the data.
No transformations will be performed yet.

```python
for name, df in datasets.items():
    print(f"\n===== {name} =====")

    column_info = pd.DataFrame({
        "column": df.columns,
        "dtype": df.dtypes.astype(str).values,
        "missing_values": df.isna().sum().values,
        "unique_values": df.nunique().values
    })

    display(column_info)
```

## 6. Data Types

Correct data types are important for the next stages of the project.

For example:

- Numeric columns should be stored as numeric values
- Identifiers may need to be treated separately from continuous variables
- Time-related columns may require special handling
- Categorical columns will later be useful for segmentation and modeling

We will not change any data types yet.
First, we want to understand the current state of the raw data.

```python
for name, df in datasets.items():
    print(f"\n===== {name} =====")
    display(df.dtypes.to_frame("data_type"))
```
